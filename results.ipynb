```python
import pandas as pd
from scipy.stats import gmean
```

```python
df_moment = pd.read_pickle('results_moment.pkl')
df_moment
```

```python
df_chronos = pd.read_pickle('chronos_results_base.pkl')
df_chronos
```

```python
df_timesfm = pd.read_pickle('timesfm_results.pkl')
df_timesfm
```

```python
seasonal = pd.read_pickle('seasonal_naive_results.pkl')
seasonal = seasonal.rename(columns={'mae':'MAE','smape':'sMAPE'})
```

```python
seasonal_result = seasonal[['dataset_name','MAE','mase','sMAPE']]
seasonal_result
```

```python
times_result = df_timesfm[['MAE','mase','sMAPE']] / seasonal[['MAE','mase','sMAPE']]
times_result
```

```python
chronos_result = df_chronos[['MAE','mase','sMAPE']] / seasonal[['MAE','mase','sMAPE']]
chronos_result
```

```python
seasonal_moment  = seasonal_result.iloc[[14,4,15,10,8,7,11,12,9]].reset_index(drop=True)
seasonal_moment
```

```python
moment_results = df_moment[['MAE','mase','sMAPE']] / seasonal_moment[['MAE','mase','sMAPE']]
moment_results
```

```python
times_result.apply(gmean)
```

```python
MAE_val = [times_result.apply(gmean)['MAE'],chronos_result.apply(gmean)['MAE'], moment_results.apply(gmean)['MAE']]


mase = [
    gmean(pd.to_numeric(df_timesfm['mase'], errors='coerce').dropna()),
    gmean(pd.to_numeric(df_chronos['mase'], errors='coerce').dropna()),
    gmean(pd.to_numeric(df_moment['mase'], errors='coerce').dropna())
]

sMAPE = [
    gmean(pd.to_numeric(df_timesfm['sMAPE'], errors='coerce').dropna()),
    gmean(pd.to_numeric(df_chronos['sMAPE'], errors='coerce').dropna()),
    gmean(pd.to_numeric(df_moment['sMAPE'], errors='coerce').dropna())
]


models = ['TimesFM-200M','Chronos-Base-200M','Moment-large-300M']
```

```python
import matplotlib.pyplot as plt

# 1. Define your 3 categories and 3 corresponding numbers


# 2. Create the bar chart
plt.bar(models, MAE_val, color=['skyblue', 'salmon', 'lightgreen'], edgecolor='black')

# 3. Add titles and labels for clarity
plt.title('Agg MAE')
plt.xlabel('Models')
plt.savefig("mae.png", format="png", dpi=300, bbox_inches="tight")
# 4. Display the chart
plt.show()
```

```python
plt.bar(models, mase, color=['skyblue', 'salmon', 'lightgreen'], edgecolor='black')

# 3. Add titles and labels for clarity
plt.title('Agg MASE')
plt.xlabel('Models')

plt.savefig("mase.png", format="png", dpi=300, bbox_inches="tight")
plt.show()

# 4. Display the chart
```

```python
plt.bar(models, sMAPE, color=['skyblue', 'salmon', 'lightgreen'], edgecolor='black')
# 3. Add titles and labels for clarity
plt.title('Agg sMAPE')
plt.xlabel('Models')
plt.savefig("smape.png", format="png", dpi=300, bbox_inches="tight")
# 4. Display the chart
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

models = ["Model A", "Model B", "Model C"]

metric1 = [85, 90, 88]
metric2 = [78, 84, 91]
metric3 = [92, 87, 89]

x = np.arange(len(models))
width = 0.25

fig, ax = plt.subplots(figsize=(10, 6))

ax.bar(x - width, metric1, width, label="Metric 1")
ax.bar(x,         metric2, width, label="Metric 2")
ax.bar(x + width, metric3, width, label="Metric 3")

ax.set_xlabel("Models")
ax.set_ylabel("Metric Value")
ax.set_title("Model Performance Comparison")
ax.set_xticks(x)
ax.set_xticklabels(models)
ax.legend()

plt.tight_layout()
plt.show()
```
