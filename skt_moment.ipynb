```python
from darts import TimeSeries, datasets

from sktime.forecasting.base import ForecastingHorizon
import warnings
from darts.utils.missing_values import fill_missing_values
from sktime.performance_metrics.forecasting import mean_absolute_error, MeanAbsoluteScaledError, mean_absolute_percentage_error
from sktime.forecasting.momentfm import MomentFMForecaster
from sklearn.preprocessing import StandardScaler


```

```python
warnings.filterwarnings('ignore')
```

```python
result_moment = []
Mase_con = MeanAbsoluteScaledError()
```

```python
#### tempreture dataset

y_nan = datasets.TemperatureDataset().load()
y = fill_missing_values(y_nan,fill="auto")    ###### HANDLE MISSING VALUES  
warnings.filterwarnings('ignore')
y_train, y_test = y.split_before(len(y)-30)
fh = ForecastingHorizon(y_test.time_index, is_relative=False)
forecaster = MomentFMForecaster(seq_len = 1095, device='cuda')

forecaster.fit(y_train.to_dataframe(), fh=list(range(1,31)))
y_pred = forecaster.predict()

mae = mean_absolute_error(y_test.to_dataframe(),y_pred)
mase = Mase_con(y_test.to_dataframe(), y_pred, y_train=y_train.to_dataframe())
smape = mean_absolute_percentage_error(y_test.to_dataframe(),y_pred, symmetric = True)

new_row = {"data": "tempreture","MAE":mae,"mase":mase,"sMAPE":smape}
result_moment.append(new_row)
```

```python
warnings.filterwarnings('ignore')
scaler = StandardScaler()
k = datasets.TaylorDataset().load()
forecaster = MomentFMForecaster(seq_len = 144, device='cuda', max_lr=0.000001)

k_train, k_test = k.split_before(len(k)- 48)
fh = ForecastingHorizon(k_test.time_index, is_relative=False)


forecaster.fit(scaler.fit_transform(k_train.to_dataframe()), fh=list(range(1,49)))
k_pred_scaled = forecaster.predict()
k_pred = scaler.inverse_transform(k_pred_scaled)

mae = mean_absolute_error(k_test.to_dataframe(),k_pred)
mase = Mase_con(k_test.to_dataframe(), k_pred, y_train=k_train.to_dataframe())
smape = mean_absolute_percentage_error(k_test.to_dataframe(),k_pred, symmetric = True)

new_row = {"data": "Taylor","MAE":mae,"mase":mase,"sMAPE":smape}
result_moment.append(new_row)
```
